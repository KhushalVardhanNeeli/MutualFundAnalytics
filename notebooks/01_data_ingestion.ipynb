# 01 — Data Ingestion

Load all 10 CSV datasets, inspect shapes, types, and anomalies.

```python
import pandas as pd
from pathlib import Path

DATA = Path('data/raw')
files = sorted(DATA.glob('*.csv'))
print(f'Found {len(files)} CSV files')
files
```

```python
summary = []
for f in files:
    df = pd.read_csv(f)
    row = {
        'file': f.name,
        'shape': df.shape,
        'columns': len(df.columns),
        'missing': df.isnull().sum().sum(),
        'duplicates': df.duplicated().sum(),
    }
    summary.append(row)
    print(f'{f.name:<40s}  shape={str(df.shape):<15s}  missing={df.isnull().sum().sum():>5d}  dup={df.duplicated().sum():>5d}')

summary_df = pd.DataFrame(summary)
summary_df
```

```python
# Deep-dive into each dataset
for f in files:
    df = pd.read_csv(f)
    print(f"\n{'='*70}")
    print(f'  {f.name}')
    print(f"{'='*70}")
    print(f'Shape: {df.shape}')
    print(f'\nColumns:')
    print(df.columns.tolist())
    print(f'\nDtypes:')
    print(df.dtypes)
    print(f'\nHead:')
    print(df.head(3))
    print(f'\nDescribe:')
    print(df.describe(include='all'))
```
